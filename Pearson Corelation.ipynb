```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr
```

```python
# Load the dataset
df = pd.read_excel("train_dataset.xlsx")  # Update path if needed

# Drop missing values in the relevant columns
clean_df = df.dropna(subset=["RFL_Att", "FSO_Att"])
```

```python
# Compute Pearson correlation
r_value, p_value = pearsonr(clean_df["RFL_Att"], clean_df["FSO_Att"])
```

```python
# Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x="RFL_Att", y="FSO_Att", data=clean_df, s=50, color='navy', edgecolor='white')
sns.regplot(x="RFL_Att", y="FSO_Att", data=clean_df, scatter=False, color='crimson', line_kws={"linewidth": 2})

# Annotate correlation value
plt.title("Pearson Correlation: RF vs FSO Attenuation", fontsize=14)
plt.xlabel("RFL_Att (RF Attenuation)")
plt.ylabel("FSO_Att (FSO Attenuation)")
plt.text(0.05, 0.95, f"r = {r_value:.4f}\np = {p_value:.2e}",
         ha='left', va='top', transform=plt.gca().transAxes,
         fontsize=12, bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Correlation coefficient (r): ≈ 0.0633

P-value: extremely small (≈ 2.36e-39), indicating statistical significance

The correlation is very weak (close to 0), even though it's statistically significant due to large sample size.

This implies that FSO and RF attenuations are not linearly related, supporting the need for non-linear models (e.g., Random Forest, Mutual Information).
